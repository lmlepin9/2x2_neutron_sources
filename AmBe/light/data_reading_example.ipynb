```python
import h5flow
import numpy as np
import matplotlib.pyplot as plt
```

```python
path = '/global/cfs/cdirs/dune/www/data/2x2/sandbox/v11/flow/'
filename = 'packet-0050017-2024_07_09_00_04_33_CDT.FLOW.hdf5' # example file, change to the one you downloaded
```

```python
import h5py
with h5py.File(path + filename, "r") as f:
    print("Keys in the HDF5 file:", list(f['charge'].keys())) # Add a key to f to go through the structure of the file e.g. f['charge]. When there are no more keys, you have reached the data. You can print the shape of the data and the data fields using .dtype.names
```

```python
file = h5flow.data.H5FlowDataManager(path + filename, "r")
```

```python
file['charge/events/', slice(0,10)] # first 10 entries in the charge/events
```

```python
file['light/events/', slice(0,10)] # first 10 entries in the light/events
```

```python
file['light/wvfm/', slice(0,10)] # first 10 entries in the light/wvfms
```

```python
# We link the charge and light events through references, h5flow will do this for us
file['charge/events', 'light/events', 'light/wvfm', slice(0,1)] # the waveforms associated to the first charge event
```

```python
# Samples has shape (8, 64, 1000) corresponding to 8 adcs, 64 channels per adc (not all used!) and 1000 samples. Each sample corresponds to a tick of 16 ns. NOTE: ADCs are not corresponding to physical location directly, TPCs are. There is a mapping below.

# Plot a waveform for one channel, e.g. adc 0, channel 4, which corresponds to TPC 0, z-side -z, bottom position (see TRAP_NAME_MAP below for the mapping of channels to TPCs and positions)
samples = file['light/wvfm/', slice(0,10)]['samples'] # first

print(np.unravel_index(np.argmax(samples, axis=None), samples.shape))# find the maximum (so we plot something interesting)

plt.plot(samples[8,5,4,:]) # plot the waveform for adc 5, channel 4
plt.xlabel('Tick (16 ns)')
plt.ylabel('ADC counts')
plt.show()

# Hint: the waveforms should be probably baseline subtracted before using them
```

```python
# one charge events: waveforms and charge hits
print(file['charge/events', 'light/events', 'light/wvfm', slice(0,1)]['samples'].shape)
print(file['charge/events', 'charge/calib_final_hits', slice(1,2)][['x', 'y', 'z']].shape)
```

```python
# plot the charge hits
hits = file['charge/events', 'charge/calib_final_hits', slice(0,100)][['x', 'y', 'z']]
plt.scatter(hits['x'], hits['z'], alpha=0.5) # top view: 100 events overlayed. You can see muon tracks crossing the modules and the division into 4 modules. Some hits are reconstructed outside the detector (which is obviously wrong)
plt.xlabel('x (cm)')
plt.ylabel('z (cm)')
plt.show()
```
